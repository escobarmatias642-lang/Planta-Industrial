```python
lisDeDatos = df['Datetime'].diff()
lisDedatos1 = df['Datetime'].to_list()
def frecuencia (list1):
  listaDeMuestreo = [None]
  for i in list1:
    resultado = (list1[i]-list1[i-1])
    listaDeMuestreo.append(resultado)
    return listaDeMuestreo

listaFrecuencia = frecuencia(lisDedatos1)
```
